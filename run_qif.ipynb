```python
from neuron_models import QIFModel, run_qif_simulation, launch_qif_app
import gradio as gr

from matplotlib import pyplot as plt
from matplotlib import rcParams
rcParams["font.family"] = "Times New Roman"
rcParams["font.size"] = 12
rcParams["axes.spines.top"] =  False
rcParams["axes.spines.right"] = False
```

## Run simulation

```python
dt = 0.01    # ms
T  = 200     # ms
iterations = int(T / dt)

model = QIFModel(tau=10, V_peak=50, V_reset=4.1, T_ref=5)

I = -16 # constant drive

time, V, spikes = model.solve(
    V0=30,
    dt=dt,
    iterations=iterations,
    I_ext=I,
    use_refractory=False
)

plt.figure(figsize=(8,4))
plt.plot(time, V)
plt.xlabel("Time (ms)")
plt.ylabel("V (mV)")
plt.title("Quadratic Integrate-and-Fire neuron")
plt.show()

print("Spike times (ms):", spikes)
```
