## RS

```python
T = 1000.0       # ms
dt = 0.1         # timestep
iters = int(T / dt)

# Build input current same length as simulation
I_ext = np.zeros(iters)
# I_ext[int(20 / dt): int(100 / dt)] = 0
# I_ext[int(100 / dt): int(200 / dt)] = 10
# I_ext[int(420 / dt): int(500 / dt)] = 0
# I_ext[int(500 / dt): int(600 / dt)] = 10
I_ext[int(100/dt):] = 10

neuron_type = "RS"
params = neuron_models[neuron_type]

# instantiate neuron
neuron = IzhikevichModel(I=I_ext, params=params)

# initial values
v0 = -65.0
u0 = params["b"] * v0

# solve
time, v_time, u_time, spike_times = neuron.solve(v0=v0, u0=u0, dt=dt, iter=iters)



fig, axes = plt.subplots(3, 1, figsize=(8, 6.5), sharex=False)
fig.tight_layout(pad=3)

ax = axes[0]
ax.plot(time, I_ext, color="black", linewidth=1.2)
ax.set_xlabel("Time (ms)", fontsize=14)
ax.set_ylabel("Iext", fontsize=14)

ax = axes[1]
ax.plot(time, v_time, color="black", linewidth=1.2)
ax.set_ylabel("v(t)", fontsize=14)
ax.set_xlabel("Time (ms)", fontsize=14)
ax = axes[2]
ax.scatter(spike_times, np.ones_like(spike_times),
           marker='|', s=200, color='black')
ax.set_xlim(0, T)

ax.set_ylim(0, 2)
ax.set_ylabel("Spikes", fontsize=14)
ax.set_xlabel("Time (ms)", fontsize=14)

plt.show()
```

## RZ

```python
T = 1000.0       # ms
dt = 0.1         # timestep
iters = int(T / dt)

# Build input current same length as simulation
# I_ext = np.zeros(iters)
# I_ext[int(0/dt):] = -5
# I_ext[int(100/dt):] = -1
# I_ext[int(500/dt):int(550/dt)] = 1
# I_ext[int(20 / dt): int(100 / dt)] = 0
# I_ext[int(100 / dt): int(200 / dt)] = 10
# I_ext[int(420 / dt): int(500 / dt)] = 0
# I_ext[int(500 / dt): int(600 / dt)] = 10

# Poission train
rate = 20.0  # spikes per second (Hz)
p_spike = rate * dt / 1000.0   # probability per timestep
I_ext = np.zeros(iters)
for i in range(iters):
    if np.random.rand() < p_spike:
        I_ext[i:i+int(5/dt)] += 1.0   # brief synaptic pulse lasting 5 ms

# Sinusoidal
time = np.linspace(0, T, iters)
f = 10  # frequency in Hz
amp = 5  # amplitude of current
offset = 1  # DC offset if needed

I_ext = amp * np.sin(2 * np.pi * f * (time / 1000)) + offset


neuron_type = "RZ"
params = neuron_models[neuron_type]

# instantiate neuron
neuron = IzhikevichModel(I=I_ext, params=params)

# initial values
v0 = -65.0
u0 = params["b"] * v0

# solve
time, v_time, u_time, spike_times = neuron.solve(v0=v0, u0=u0, dt=dt, iter=iters)



fig, axes = plt.subplots(3, 1, figsize=(8, 6.5), sharex=False)
fig.tight_layout(pad=3)

ax = axes[0]
ax.plot(time, I_ext, color="black", linewidth=1.2)
ax.set_xlabel("Time (ms)", fontsize=14)
ax.set_ylabel("Iext", fontsize=14)

ax = axes[1]
ax.plot(time, v_time, color="black", linewidth=1.2)
ax.set_ylabel("v(t)", fontsize=14)
ax.set_xlabel("Time (ms)", fontsize=14)
ax = axes[2]
ax.scatter(spike_times, np.ones_like(spike_times),
           marker='|', s=200, color='black')
ax.set_xlim(0, T)

ax.set_ylim(0, 2)
ax.set_ylabel("Spikes", fontsize=14)
ax.set_xlabel("Time (ms)", fontsize=14)

plt.show()
```
