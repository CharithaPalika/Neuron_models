```python
from neuron_models import FNModel, launch_FN_app
import numpy as np
from matplotlib import pyplot as plt
from scipy import signal
from tqdm.autonotebook import tqdm
from matplotlib import rcParams
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['DejaVu Serif']
rcParams["font.size"] = 12
rcParams["axes.spines.top"] =  False
rcParams["axes.spines.right"] = False
```

## Run simulation

```python
# Parameters
v0 = 0.3
w0 = 0
Iext = 0.6
model = FNModel(a = 0.5,
                b = 0.1,
                r = 0.1,
                Iext = Iext)

# Solving ODE
t = np.arange(0,100,0.01)
v_time, w_time, dv, dw = model.solve( v0 = v0, w0 = w0, dt = 0.01, iter = 10000)

# Plotting
plt.figure(figsize = (12,4))

plt.subplot(1,3,1)
plt.title('(a) \n Voltage and w', fontsize = 10)
plt.xlabel('time (ms)', fontsize = 10)
plt.ylabel('Voltage (mV)', fontsize = 10)
plt.plot(t,v_time, label = 'voltage')
plt.plot(t,w_time, label = 'w')
plt.legend(fontsize = 8)

# Nullclines
plt.subplot(1,3,2)
plt.plot(v_time,
         w_time,
         linestyle="-",
         label = 'Trajectory',
         color = 'mediumseagreen')

plt.quiver(v_time[100],
           w_time[100],
           dv[100], dw[100], color = 'darkgreen')
          #  headwidth = 9,
          #  headlength = 7,
          #  color = 'darkgreen',
          #  scale = 4)

plt.quiver(v_time[500],
           w_time[500],
           dv[500], dw[500], color = 'darkgreen')
          #  headwidth = 9,
          #  headlength = 7,
          #  color = 'darkgreen',
          #  scale = 3)

plt.legend(fontsize = 8)
plt.title('(b) \n Trajectory' ,fontsize = 10)
plt.xlabel('v', fontsize = 10)
plt.ylabel('w', fontsize = 10)


# Phase plot
plt.subplot(1,3,3)
startpoints = None
density = 0.3
v_range = np.linspace(-0.5,1.5,100)
v0 = np.linspace(-0.5,1.5,5)
w0 = np.linspace(-0.5,1.5,5)
model.phaseplot(v0, w0, startpoints, density)
plt.title('(b) \n Phase plane' ,fontsize = 10)
plt.xlabel('v', fontsize = 10)
plt.ylabel('w', fontsize = 10)

plt.tight_layout()
plt.show()
```
